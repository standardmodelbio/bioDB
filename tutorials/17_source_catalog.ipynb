````python
import gzip
import tempfile
from pathlib import Path

import pandas as pd

work = Path(tempfile.mkdtemp(prefix="biodb-tutorial-"))
````

````python
from biodb import aou_allxall, gtr, omicspred, panukbb

burden = pd.DataFrame(
    {
        "gene_id": ["ENSG1", "ENSG2"],
        "gene_symbol": ["GENEA", "GENEB"],
        "analysis_id": ["100001", "100001"],
        "ancestry_group": ["meta", "meta"],
        "annotation": ["pLoF", "missenseLC"],
        "max_maf": [0.001, 0.001],
        "neg_log10_p_burden": [10.0, 6.0],
        "neg_log10_p_skat": [8.0, 4.0],
        "neg_log10_p": [9.0, 5.0],
        "beta_burden": [2.3, -1.2],
    }
)
long = aou_allxall.melt_gene_burden(burden, test="burden", beta_signed=True)
assert long["score"].tolist() == [10.0, -6.0]
print(long)
# SKAT/SKAT-O significance belongs to different tests; beta is the burden effect.
print(aou_allxall.melt_gene_burden(burden, test="skat", beta_signed=False))
manifest = pd.DataFrame(
    {
        "phenocode": ["traitA", "traitB", "traitC"],
        "trait_type": ["biomarkers"] * 3,
        "sldsc_25bin_h2_z_EUR": [9, 8, 3],
        "phenotype_qc_EUR": ["PASS"] * 3,
        "category": ["blood"] * 3,
        "lambda_gc_EUR": [1.5, 1.4, 1.0],
    }
)
selected = panukbb.select_traits(manifest, n=5, max_per_category=1)
assert selected["phenocode"].tolist() == ["traitA"]
print(selected)
````

The All-by-All table's mask × MAF × ancestry × test axes are separate analysis
choices. A signed burden-score transform is a presentation choice; raw GWAS
acquisition in [13](13_raw_gwas.ipynb) preserves measured fields instead.
Pan-UKBB `select_traits` defaults to QC PASS, suitable trait types and
heritability Z > 4. Genomic-inflation filtering is opt-in; it is not a universal
confounding threshold for highly polygenic traits.


````python
score_file = work / "score.txt.gz"
score_file.write_bytes(
    gzip.compress(
        b"##pgs_id=synthetic\nrsID\tchr_name\tchr_position\teffect_allele\tother_allele\teffect_weight\nrs1\t1\t101\tG\tA\t0.25\n"
    )
)
print(omicspred.read_scoring_file(score_file))
# A GTR lightweight TSV cache feeds real per-test sets, deduplication and GMT.
rows = [
    ["GTR0001.1", "clinical", "condition", "C1", "", "Synthetic condition", "", ""],
    ["GTR0001.1", "clinical", "gene", "", "", "GENEA", "1", "GENEA"],
    ["GTR0002.1", "clinical", "condition", "C1", "", "Synthetic condition", "", ""],
    ["GTR0002.1", "clinical", "gene", "", "", "GENEA", "1", "GENEA"],
]
pd.DataFrame(
    rows,
    columns=[
        "#accession_version",
        "test_type",
        "object",
        "GTR_identifier",
        "MIM_number",
        "object_name",
        "gene_or_SNOMED_CT_ID",
        "gene_symbol",
    ],
).to_csv(work / gtr.TEST_CONDITION_GENE_FILE, sep="\t", index=False)
sets = gtr.aggregate_gene_sets(by="condition", cache_dir=work)
assert sets["support_count"].tolist() == [2]
print(sets)
print(
    gtr.panel_text(
        {
            "name": "Synthetic panel",
            "conditions": [{"name": "Synthetic condition"}],
            "methods": ["Sequencing"],
            "clinical_utility": "Synthetic text only",
        }
    )
)
gtr.to_gmt(work / "panels.gmt", by="condition", cache_dir=work)
print((work / "panels.gmt").read_text().strip())
````
